```python
import spacy #object oriented ,very user friendly
```

```python
nlp=spacy.load("en_core_web_sm")

doc=nlp("The Number of tasks etc. were 45. It was done by Hulk")
for sentence in doc.sents:
  print(sentence)
```

```python
for sentence in doc.sents:
  for word in sentence:
    print(word)
```

```python
with open("students (1).txt", "r", encoding="utf-8") as file:
    text = file.readlines()

print(text)
```

```python
text = " ".join(text) # use only with readlines() otherwise will give empty list
text
```

```python
doc = nlp(text)

emails = []

for token in doc:
    if token.like_email:
        emails.append(token.text)

print("Emails found:")
print(emails)
```

Other **Language**

```python
nlp = spacy.blank("hi")
doc = nlp("भैया जी! 5000 ₹ उधार थे वो वापस देदो")
for token in doc:
    print(token, token.is_currency)
```
