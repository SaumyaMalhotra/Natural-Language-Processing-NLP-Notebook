```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

```python
doc=nlp("Elon flew to mars yesterday. He carried biryani masala with him")
for token in doc:
  print(token,"|",
        token.pos_)
```

```python
for token in doc:
  print(token,"|",
        spacy.explain(token.pos_))
```

```python
for token in doc:
  print(token,"|",
        token.tag_,"|",
        spacy.explain(token.tag_)) #spacy.explain()  -> explanation of function
```

## **NLP Analysis on a Text**

```python
text='''
   Last    weekend,   I visited   Jaipur with my friends!!!   The weather was AMAZING,  and we explored Hawa Mahal,   City Palace, and Amber Fort.

We spent around  ₹5,500 during the trip,   clicked 250+ photos, and tried delicious Rajasthani food.  Honestly,   I couldn't stop smiling :)   Although the streets were a bit crowded, the experience was absolutely worth it!!!

After returning home, I uploaded the pictures to Instagram,  shared them with my family, and wrote,  "Jaipur is one of the most beautiful cities I've ever visited."     Overall,   it was a memorable trip, and I'd love to visit again in 2027.     #Travel #Jaipur @friends
'''
```

```python
doc=nlp(text)
```

```python
for token in doc:
    print(token,"|",
          spacy.explain(token.pos_))
```

```python
filtered_tokens=[]

for token in doc:
  if token.pos_ not in ["SPACE","X","PUNCT"]:
    filtered_tokens.append(token)
```

```python
filtered_tokens #clean tokens
```

```python
count=doc.count_by(spacy.attrs.POS)
count
```

```python
doc.vocab[92].text
```

```python
for k , v in count.items():
  print(doc.vocab[k].text, ":" ,v)
```

# **EXERCISE**

You are parsing a news story from cnbc.com. News story is stores in news_story.txt which is available in this same folder on github. You need to,

1. Extract all NOUN tokens from this story. You will have to read the file in python first to collect all the text and then extract NOUNs in a python list
2. Extract all numbers (NUM POS type) in a python list
3. Print a count of all POS tags in this story

```python
from google.colab import files
uploaded = files.upload()
```

```python
with open ("news_story.txt","r") as f:
  text=f.readlines()
  print(text)
```

```python
text = " ".join(text) # use only with readlines() otherwise will give empty list
text
```

```python
doc=nlp(text)
for ntoken in doc:
  print(ntoken , "|" ,ntoken.pos_)
```

```python
for token in doc:
    if token.is_punct:
        print(token.text, "|", token.pos_)
```

```python
for token in doc:
    if token.is_punct or token.is_space:
        continue
    print(token.text, "|", token.pos_)
```

```python
noun_text=[]
for token in doc:
    if token.pos_ == "NOUN" and token.text != "%":
      noun_text.append(token.text)
```

```python
noun_text[:10]
```

```python
num_tokens=[]
for tokens in doc:
  if tokens.pos_=="NUM":
    num_tokens.append(tokens)
```

```python
num_tokens
```

```python
count = doc.count_by(spacy.attrs.POS)
```

```python
for k,v in count.items():
  print(doc.vocab[k].text, ":" ,v)
```
